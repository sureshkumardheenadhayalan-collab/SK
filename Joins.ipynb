```python
#InnerJoin

import sqlite3

# Connect to database
conn = sqlite3.connect("C:\\Users\\LENOVO\\OneDrive\\Desktop\\files\\retail.db")

# Create cursor
cur = conn.cursor()

# Execute SQL query
cur.execute("""
SELECT o.order_id, o.order_date, o.status,
       c.name AS customer, c.country
FROM orders AS o
JOIN customers AS c
    ON c.customer_id = o.customer_id
ORDER BY o.order_id
LIMIT 5
""")

# Fetch records
rows = cur.fetchall()

# Display column headers
#print("Order ID | Order Date | Status | Customer | Country")
#print("-" * 65)

# Display records
for row in rows:
    print(row)

# Close connection
conn.close()

```

```python
#left Join


import sqlite3

# Connect to database
conn = sqlite3.connect("C:\\Users\\LENOVO\\OneDrive\\Desktop\\files\\retail.db")

# Create cursor
cur = conn.cursor()

# Execute SQL query
cur.execute("""
SELECT c.customer_id, c.name, COUNT(o.order_id) AS orders
FROM customers AS c
LEFT JOIN orders AS o
    ON o.customer_id = c.customer_id
GROUP BY c.customer_id, c.name
ORDER BY orders, c.customer_id
LIMIT 6
""")

# Fetch records
rows = cur.fetchall()

# Display column headers
print("Customer ID | Name | Orders")
print("-" * 40)

# Display records
for row in rows:
    print(row)

# Close connection
conn.close()
```

```python
# Anti Join


import sqlite3

# Connect to database
conn = sqlite3.connect("C:\\Users\\LENOVO\\OneDrive\\Desktop\\files\\retail.db")

# Create cursor
cur = conn.cursor()

# Execute SQL query
cur.execute("""
SELECT c.country, COUNT(*) AS customers_without_orders
FROM customers AS c
LEFT JOIN orders AS o
    ON o.customer_id = c.customer_id
WHERE o.order_id IS NULL
GROUP BY c.country
ORDER BY customers_without_orders DESC
""")

# Fetch records
rows = cur.fetchall()

# Display column headers
print("Country | Customers Without Orders")
print("-" * 40)

# Display records
for row in rows:
    print(row)

# Close connection
conn.close()
```

```python
#Having


import sqlite3

# Connect to database
conn = sqlite3.connect("C:\\Users\\LENOVO\\OneDrive\\Desktop\\files\\retail.db")

# Create cursor
cur = conn.cursor()

# Execute SQL query
cur.execute("""
SELECT 'filter in ON' AS version,
       COUNT(DISTINCT c.customer_id) AS customers_kept
FROM customers AS c
LEFT JOIN orders AS o
    ON o.customer_id = c.customer_id
    AND o.status = 'returned'

UNION ALL

SELECT 'filter in WHERE',
       COUNT(DISTINCT c.customer_id)
FROM customers AS c
LEFT JOIN orders AS o
    ON o.customer_id = c.customer_id
WHERE o.status = 'returned'
""")

# Fetch records
rows = cur.fetchall()

# Display column headers
print("Version | Customers Kept")
print("-" * 40)

# Display records
for row in rows:
    print(row)

# Close connection
conn.close()
```

```python
#Revenue
import sqlite3
import pandas as pd

conn = sqlite3.connect("C:\\Users\\LENOVO\\OneDrive\\Desktop\\files\\retail.db")

def sql(query, params=()):
    return pd.read_sql_query(query, conn, params=params)

print("SQLite version:", sqlite3.sqlite_version)

# Get all table names
tables = sql(
    "SELECT name FROM sqlite_master WHERE type='table' ORDER BY name"
)["name"]

# print(tables)

# Display table name, row count, and column names
df = pd.DataFrame({
    "table": tables,
    "rows": [
        conn.execute(f'SELECT COUNT(*) FROM "{t}"').fetchone()[0]
        for t in tables
    ],
    "columns": [
        ",".join(
            row[1] for row in conn.execute(f'PRAGMA table_info("{t}")')
        )
        for t in tables
    ],
})

# print(df)

# customers = sql("SELECT * FROM customers limit 5")
# print(customers.to_string(index=False))

# customers = sql("SELECT * FROM orders limit 5")
# print(customers.to_string(index=False))

demo = sql("""
    SELECT
		strftime('%Y-%m', o.order_date) AS order_month,
		COUNT(DISTINCT o.order_id) AS total_orders,
		SUM(oi.quantity * oi.unit_price) AS total_revenue
	FROM orders o
	JOIN order_items oi
		ON o.order_id = oi.order_id
	GROUP BY strftime('%Y-%m', o.order_date)
	ORDER BY order_month;
""")

print(demo.to_string(index=False))
conn.close()
```
